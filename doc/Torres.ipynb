# Torres de Hanoi

El estado es una tupla de tres torres; cada torre es una tupla de discos ordenados
de menor a mayor. Mover un disco cuesta 1, asi que $g(n)$ coincide con la profundidad.
El numero minimo de movimientos para $n$ discos es $2^n-1$.


```python
import sys,os
baile_path=os.path.abspath(os.path.join("..","src"))
if not baile_path in sys.path:
    sys.path.append(baile_path)
import SimpleSearch as ss
```

```python
inicio=ss.node(((1,2,3,4,5,6,7),(), ()), op='start') # estado inicial
final=ss.node(((),(), (1,2,3,4,5,6,7)))                # estado meta
```

```python
def meta(*estados):
    inicio=estados[0]
    final=estados[1]
    return inicio.state==final.state
```

```python
def mover(nodo):
    T=nodo.state
    sucesores=[]
    for o,to in enumerate(T):
        for d, td in enumerate(T):
            if o!=d and  len(to)>0:
                if len(td)==0 or to[0]<td[0]:
                    n=[list(t) for t in T]
                    n[o]=to[1:]
                    n[d]=to[:1]+td
                    s=tuple([tuple(t) for t in n])
                    nn=ss.node(s, depth=nodo.depth+1, 
                               op=f'mover de {o} a {d}', parent=nodo)
                    sucesores.append(nn)
    return sucesores
```

```python
bfs=ss.TreeSearch(inicio,mover,meta, goal_state=final, strategy='bfs')
```

```python
r=bfs.find()
```

```python
dfs=ss.TreeSearch(inicio,mover,meta, goal_state=final, strategy='dfs')
```

```python
def h(s,f):
    return len(s.state[0])+len(s.state[1])
```

```python
astar=ss.TreeSearch(inicio,mover,meta, goal_state=final, strategy='a*', heuristic=h)
```

```python
ra=astar.find()
```

```python
len(ra.getPath())
```

### Comparacion de las tres estrategias

Cada busqueda guarda en **iterations** el numero de nodos expandidos, y el nodo
solucion guarda en **cost** el costo acumulado $g(n)$ (aqui, un movimiento por paso).


```python
for nombre,busqueda in [('BFS',bfs),('DFS',dfs),('A*',astar)]:
    r=busqueda.find()
    print('%-4s expandidos: %6d   longitud de la solucion: %3d'
          %(nombre, busqueda.iterations, r.cost))
```
